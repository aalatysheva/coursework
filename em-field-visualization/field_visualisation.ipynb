```python
import numpy as np
import matplotlib.pyplot as plt

# Константы
k_e = 1  # Коэффициент пропорциональности (безразмерный)
q = 1    # Модуль зарядов (безразмерный)

# Параметры сетки
x = np.linspace(-3, 3, 100)
y = np.linspace(-3, 3, 100)
X, Y = np.meshgrid(x, y)

# Параметры зарядов (4 заряда в вершинах квадрата)
charges = [(1, 1, 1), (1, -1, -1), (-1, -1, 1), (-2, 2, -2),(-1,1,3) ]  # (x, y, q)

# Расчет электрического потенциала
V = np.zeros(X.shape)

for charge in charges:
    qx, qy, q_value = charge
    r = np.sqrt((X - qx)**2 + (Y - qy)**2)
    V += k_e * q_value / r  # Сумма потенциалов от всех зарядов

# Построение силовых линий
Ex, Ey = np.zeros(X.shape), np.zeros(Y.shape)

for charge in charges:
    qx, qy, q_value = charge
    r = np.sqrt((X - qx)**2 + (Y - qy)**2)
    Ex += k_e * q_value * (X - qx) / r**3
    Ey += k_e * q_value * (Y - qy) / r**3

# Построение графиков на одном рисунке
plt.figure(figsize=(8, 8))

# Силовые линии
plt.streamplot(X, Y, Ex, Ey, color='b', density=2, linewidth=1.5, arrowsize=1.5)

# Эквипотенциальные линии
contour = plt.contour(X, Y, V, levels=150, colors='r', alpha=0.7)


plt.title('Силовые линии и эквипотенциальные линии электрического поля')
plt.xlabel('X')
plt.ylabel('Y')
plt.grid()
plt.axis('auto')
plt.tight_layout()
plt.show()
```
